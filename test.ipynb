```python
import polars as pl
```

```python
df = pl.read_csv("../TradingAgents/results/event_studies/event_study_metrics.csv")
```

```python
df
```

```python
df = df.with_columns(
    pl.col("Return").str.replace("%", "").cast(pl.Float64),
    pl.col("Max Drawdown").str.replace("%", "").cast(pl.Float64),
    pl.col("CAPM Alpha (Ann)").str.replace("%", "").cast(pl.Float64),
    pl.col("FF3 Alpha (Ann)").str.replace("%", "").cast(pl.Float64),
    pl.col("FF5 Alpha (Ann)").str.replace("%", "").cast(pl.Float64)
)
```

```python

df.group_by("Strategy").agg(pl.col("Return").mean()).sort("Return", descending=True)
```
